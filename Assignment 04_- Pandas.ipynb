## Question 1: Data Loading and Basic Information
**Problem Statement:**
1. Load a CSV file named "data.csv" into a Pandas DataFrame.
2. Display the first 5 rows of the DataFrame.
3. Display the column names and data types in the DataFrame.
4. Show the summary statistics (mean, median, min, max, etc.) for numerical columns.

```python
import pandas as pd
df=pd.read_csv("Online Shopping Dataset.csv")
print(df.head())
print(df.columns)
print(df.dtypes)
print(df.describe())
```

## Question 2: Data Manipulation and Filtering
**Problem Statement:**
1. Add a new column 'Total' that contains the sum of two existing numerical columns.
2. Filter the DataFrame to show rows where a specific column value meets a certain condition (e.g., where Avg_Price > 120).
3. Sort the DataFrame based on a particular column in ascending order.

```python
import numpy as np
df["Total"]=df["Avg_Price"]+df["Discount_pct"]
print(df)
print("#"*30)
print(df[df["Avg_Price"]>120])
print("#"*30)
df_sort=df.sort_values(by="Total",ascending=True)
print(df_sort)


```

## Question 3: Statistical Analysis and Grouping
**Problem Statement:**
1. Calculate the average value of a specific column.
2. Find the most frequent value in a categorical column.
3. Group the data by a categorical column and calculate the mean for each group.

```python
average_value=df["Total"].mean()
print(average_value)
print(df["Location"].mode()[0])
df_group=df.groupby("Location")["Total"].mean()
print(df_group)



```

## Question 4: Data Visualization
**Problem Statement:**
1. Create a histogram to visualize the distribution of a numerical column.
2. Generate a bar chart to represent the frequency of categories in a categorical column.

```python
import matplotlib.pyplot as plt
df["Total"].hist() 
plt.title("Distribution")
plt.show()
a=df["Location"].value_counts()
plt.bar(a.index,a.values,color="green")
plt.title("Frequensy of Locations")
plt.show()                                  
```

## Question 5: Data Cleaning
**Problem Statement:**
1. Perform data imputation to handle missing values in the DataFrame.

```python

num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].mean())


cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum()
```
