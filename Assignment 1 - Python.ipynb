## Problem 1: Maximum and Minimum of Three Numbers

Write a program to find a maximum and a minimum of three numbers or if the numbers are equal to each other, find if the given number is positive or negative.

**Examples:**
- Input: 6, 2, 8 → Output: minimum is 2 and maximum 8
- Input: 1, 1, 1 → Output: the num is 1

````python
x,y,z=input("enter 3 numbers").split()
a=int(x)
b=int(y)
c=int(z)

if a == b == c:
    if a > 0:
        print(f"The number is {a} and it is positive.")
    elif a < 0:
        print(f"The number is {a} and it is negative.")
    else:
        print("The number is 0.")
else:
    minimum = min(a, b, c)
    maximum = max(a, b, c)
    print(f"Minimum is {minimum} and maximum is {maximum}")
````

## Problem 6: String Case Conversion

Given a string, write a python program that convert the string based on the following conditions:
- If lower case letter less than upper case letter convert string to upper case
  - Example: AAAa → AAAA
- If upper case letter less than lower case letter convert string to lower case
  - Example: aaaA → aaaa

````python
s = input("Enter the string: ")
upper_count = 0
lower_count = 0
for i in s:
    if i.isupper():
        upper_count += 1
    elif i.islower():
        lower_count += 1
if lower_count < upper_count:
    print(s.upper())
elif upper_count < lower_count:
    print(s.lower())
else:
    print(s)
````

## Problem 7: Palindrome and Armstrong Number Check

Write a program to accept number from the user and check if the given number is palindrome or Armstrong or not.

**Examples:**
- Palindrome: 12321 (reads the same forwards and backwards)
- Armstrong: 153 (1³ + 5³ + 3³ = 1 + 125 + 27 = 153)

````python
number = input("Enter a number: ")
if number.isdigit():
    num = int(number)
    if number == number[::-1]:
        print("The number is a Palindrome.")
    else:
        total = 0
        power = len(number)
        for i in number:
            total += int(i) ** power
        if total == num:
            print("The number is an Armstrong number")
        else:
            print("The number is neither Palindrome nor Armstrong")

else:
    print("Invalid input. Please enter digits only.")
````

## Problem 8: Number to Words Conversion (Bonus)

Write a program to accept number from the user and convert it to words.

**Example:**
- Input: 124 → Output: One hundred and twenty-four

````python
ones = ["", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine"]
teens = ["ten", "eleven", "twelve", "thirteen", "fourteen", "fifteen",
         "sixteen", "seventeen", "eighteen", "nineteen"]
tens = ["", "", "twenty", "thirty", "forty", "fifty", "sixty", "seventy", "eighty", "ninety"]

number = int(input("Enter a number "))

if number == 0:
    print("zero")
elif number < 10:
    print(ones[number])
elif number < 20:
    print(teens[number - 10])
elif number < 100:
    if number % 10 == 0:
        print(tens[number // 10])
    else:
        print(tens[number // 10] + "-" + ones[number % 10])
else:
    hundred = number // 100
    remainder = number % 100

    result = ones[hundred] + " hundred"

    if remainder != 0:
        if remainder < 10:
            result += " and " + ones[remainder]
        elif remainder < 20:
            result += " and " + teens[remainder - 10]
        else:
            if remainder % 10 == 0:
                result += " and " + tens[remainder // 10]
            else:
                result += " and " + tens[remainder // 10] + "-" + ones[remainder % 10]

    print(result)
````

## Problem 9: Series Sum Calculation

Write a program to find the sum of the following series: 1 + 4 – 9 +16 – 25 + 36 ...........n terms

Note: The input is n (number of terms)

**Example:**
- Input: 5 → Output: 1 + 4 – 9 +16 – 25 = -13

````python
n = int(input("Enter the number  "))

sum = 0

for i in range(1, n + 1):
    square_num= i * i

    if i % 3 == 0 or i % 2 == 1 and i != 1:
        sum -= square_num
    else:
        sum += square_num

print("Sum of the series:", sum)
````

## Problem 12: Average of Tuple of Tuples

Write a Python program to calculate the average value of the numbers in a given tuple of tuples.

**Original Tuple:**
```
((10, 10, 10, 12), (30, 45, 56, 45), (81, 80, 39, 32), (1, 2, 3, 4))
```

**Expected Output:**
```
Average value of the numbers of the said tuple of tuples: [30.5, 34.25, 27.0, 23.25]
```

````python
tuble = ((10, 10, 10, 12),
         (30, 45, 56, 45),
         (81, 80, 39, 32),
         (1, 2, 3, 4))

columns = len(tuble[0])#4
rows = len(tuble)#4
average = []

for i in range(columns):
    sum_column = 0  
    for j in range(rows):
        sum_column += tuble[j][i]  
    avr = sum_column / rows
    average.append(avr)

print(average)
````

## Problem 17: Student Dictionary Management

Write a Python program to create a dictionary that stores the names of 3 students, each student has another dictionary with their 'age' and 'grades'. Allow editing of a student's age or grades. The user should choose which student to edit, and then either update the age or modify the list of grades.

**Example Dictionary Structure:**
```python
students = {
    'Alice': {'age': 20, 'grades': [85, 90, 92]},
    'Bob': {'age': 21, 'grades': [75, 80, 78]},
    'Charlie': {'age': 19, 'grades': [95, 97, 93]}
}
```

````python
students = {
    'Alice': {'age': 20, 'grades': [85, 90, 92]},
    'Bob': {'age': 21, 'grades': [75, 80, 78]},
    'Charlie': {'age': 19, 'grades': [95, 97, 93]}
}


name = input("Enter the name of the student to edit: ")
if name in students:
    choice = input("Do you want to edit 'age' or 'grades'? ").lower()

    if choice == 'age':
        new_age = int(input("Enter the new age: "))
        students[name]['age'] = new_age
       

    elif choice == 'grades':
        print("Current grades:", students[name]['grades'])
        new_grades = []
        num = int(input("How many new grades? "))
        for i in range(num):
            grade = int(input(f"Enter grade {i+1}: "))
            new_grades.append(grade)
        students[name]['grades'] = new_grades
        

    else:
        print("Invalid choice.")
else:
    print("Student not found.")
print("\nFinal student data:")
for student, info in students.items():
    print(f"{student} → Age: {info['age']}, Grades: {info['grades']}")
````
