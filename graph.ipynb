# bar plot

```python
a=np.array(["c1","c2","c3"])
b=np.array([2,4,6])
plt.bar(a,b,color="red")
plt.xlabel("x lab")
plt.ylabel("y lab")
plt.show()
```

```python
a=np.array(["c1","c2","c3"])
b=np.array([2,4,6])
plt.barh(a,b,color="red")
plt.xlabel("y lab")
plt.ylabel("x lab")
plt.show()
```

# pie plot

```python
z=np.array([40,30,30])
r=np.array(["UI","Web","AI"])
plt.pie(z,labels=r,colors=["red","yellow","black"])
plt.show()
```

# sub plots

```python

plt.subplot(3,1,1)
plt.plot(x)
plt.subplot(3,1,2)
plt.plot(x,x**2)
plt.subplot(3,1,3)
plt.plot(x,np.sin(x))

plt.show()
```

```python
fig,axis=plt.subplots(3,1)
axis[0].plot(x)
axis[1].pie(z)
axis[2].bar(a,b)
```

# subplots

```python
fig,axis=plt.subplots(2,3)
fig.suptitle("figures",color="red")
axis[0,0].plot(x)
axis[0,1].bar(a,b)
axis[0,1].set_title("Hello world")

axis[0,2].pie(z)
axis[1,0].plot(np.sin(x))
axis[1,1].pie(z)
axis[1,1].set_title("Hello ")
axis[1,2].plot(x**2)
```

```python
fig,axis=plt.subplots(8,1,figsize=(6,4))
axis[0].plot(x)

axis[1].bar(a,b)

axis[2].pie(z)
axis[3].plot(np.sin(x))
axis[4].plot(np.sin(x))
axis[5].plot(np.cos(x))

axis[6].barh(a,b)

axis[7].plot(x)
axis[7].plot(y)
plt.show()
```
