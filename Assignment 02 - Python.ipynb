2.	Write function area() that calculates the area of the shape. (Any 2D shape ---- more than two shapes) 

```python
def area(shape, x, y=None):
    if shape == "square":
        return x * x
    elif shape == "rectangle":
        return x * y
    elif shape == "triangle":
        return 0.5 * x * y
    else:
        return None  

shape = input("Select shape (square, triangle, rectangle)").lower()

if shape in ("rectangle", "triangle"):
    x = float(input("Enter first side (positive): "))
    y = float(input("Enter second side (positive): "))
    if x > 0 and y > 0:
        print("Area:", area(shape, x, y))
    else:
        print("sides must be positive numbers.")
elif shape == "square":
    x = float(input("Enter side (positive): "))
    if x > 0:
        print("Area:", area(shape, x))
    else:
        print("side must be a positive number.")
else:
    print("Shape is not in the selectors")
```

6.	Write a function which will be able to print a index of list element without using index function.

```python
def find(lst, element):
    pos = 0
    for item in lst:
        if item == element:
            return pos
        pos += 1
    return -1  



user_list = input("enter elements with space : ").split()

user_list = [int(x) for x in user_list]


element = int(input("enter the element: "))

index = find(user_list, element)
if index != -1:
    print("Index is", index)
else:
    print("element isnot in list")
```

8.	Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order. (Note: Use dictionaries and lists to store the data.)
Perform the following tasks now:
-	Now add items to the menu.
-	Make table reservations.
-	Take customer orders.
-	Print the menu.
-	Print table reservations.
-	Print customer orders.


```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}          
        self.booked_tables = []       
        self.customer_orders = {}    

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price
        print(f"Added {item} to the menu with price {price} EGP.")

    def book_table(self, table_number):
        if table_number not in self.booked_tables:
            self.booked_tables.append(table_number)
            print(f"Table {table_number} booked successfully.")
        else:
            print(f"Table {table_number} is already booked.")

    def customer_order(self, table_number, order_list):
        if table_number in self.booked_tables:
            self.customer_orders[table_number] = order_list
            print(f"Orders for table {table_number} have been recorded.")
        else:
            print(f"Cannot record orders, Table {table_number} is not booked.")

    def print_menu(self):
        print("\n Menu:")
        for item, price in self.menu_items.items():
            print(f"- {item}: {price} EGP")

    def print_bookings(self):
        print("\n Booked tables:")
        if self.booked_tables:
            print(", ".join(map(str, self.booked_tables)))
        else:
            print("No bookings.")

    def print_orders(self):
        print("\n Customer Orders:")
        if self.customer_orders:
            for table, orders in self.customer_orders.items():
                print(f"Table {table}: {', '.join(orders)}")
        else:
            print("No orders.")


rest = Restaurant()

rest.add_item_to_menu("Pizza", 120)
rest.add_item_to_menu("Burger", 80)
rest.add_item_to_menu("Pasta", 100)

rest.book_table(1)
rest.book_table(2)

rest.customer_order(1, ["Pizza", "Pasta"])
rest.customer_order(2, ["Burger"])

rest.print_menu()
rest.print_bookings()
rest.print_orders()
```

-	**requestBike**: Takes a request from the customer for the number of bikes and check if the requested number is available or not if yes, function will deduct number_of_bikes from shop stock else it will print wait 10 mins. And you have to make sure that the age of the customer is over than 6 years. 
-	**returnBike**: Allows customers to return their bikes to the rental shop by taking rentalTime, num_of_bikes as parameters. And calculate the total bill cost by multiplying rentalTime * 40. Also, add the returned bikes to the stock store again 
-	**addBike**: adds bikes into the list 
-	**totalCost**: returns the total cost of bikes in the store 


```python
class MainBikeRental:
    def __init__(self,lst_bikes=None,lst_rentals=None):
        if lst_bikes is None:
            lst_bikes = []
        if lst_rentals is None:
            lst_rentals = []
        self.lst_bikes=lst_bikes
        self.lst_rentals=lst_rentals
    def set_lst_bikes(self,lst_bikes):
        self.lst_bikes=lst_bikes
    def set_lst_rentals(self,lst_rentals):
        self.lst_rentals=lst_rentals
    def get_lst_rentals(self):
        return self.lst_rentals
    def get_lst_bikes(self):
        return self.lst_bikes
    def requestBike(self, n_bike, age):
        if age > 6:
            if n_bike <= len(self.lst_bikes):
                for i in range(n_bike):
                    self.lst_bikes.pop(0) 
                print(f"{n_bike} bike rented successfully.")
            else:
                print("wait 10 mins")
        else:
             print("age must be > 6")

    def returnBike(self,num_of_bikes,rentalTime):
        bill = rentalTime * 40
        for i in range(num_of_bikes):
            self.lst_bikes.append(f"bike_{len(self.lst_bikes)+1}")
        print(f"Total bill: {bill} EGP")
        return bill
    def addBike(self, bikes):
        self.lst_bikes.extend(bikes)
        print(f"{len(bikes)} bike added to stock.")

    
    def totalCost(self):
        return len(self.lst_bikes) * 40
mr = MainBikeRental(["bike1", "bike2", "bike3", "bike4", "bike5"])

print( mr.get_lst_bikes())


mr.requestBike(2, 10)
print("Bikes after renting:", mr.get_lst_bikes())
print("Rentals list:", mr.get_lst_rentals())

print("#"*20)
mr.returnBike(1, 3)
print("Bikes after returning:", mr.get_lst_bikes())
print("#"*20)

mr.addBike(["bike6", "bike7"])

print("Bikes after adding:", mr.get_lst_bikes())
print("#"*20)
print("Total stock value:", mr.totalCost(), "EGP")
```
