## Question 4

Write a NumPy program to find the most frequent value in an array.

```python
import numpy as np
from scipy import stats

arr = np.array([1, 3, 2, 5, 2, 3, 5, 4, 1, 3,5,6,5])

value = stats.mode(arr, keepdims=True)


print("Most frequent value is", value.mode[0])
```

## Question 7

Write a NumPy program to create a 3x3 array with random values and subtract the mean of each row from each element.

```python
from numpy import *
my_array=np.random.random((3,3))
print(my_array)
for i in my_array:
    res=subtract(i,mean(i))
    print(f"the mean of each row is {res}")
```

## Question 9

Write a NumPy program to create a 5x5 array with random values and find the second-largest value in each column.

```python
import numpy as np
arr=np.random.randint(1,100,size=(5,5))
sorted=np.sort(arr,axis=0)
print(sorted)
second=sorted[-2,:]
print("Second largest value in each column:", second)
```
